```python
import numpy as np

x_train=np.array([1,2,4,5])
y_train=np.array([50,55,70,75])
m=x_train.shape[0]

models=[(5,45),(8,40)]

for w,b in models:
    f=w*x_train+b
    errors=f-y_train
    squared=errors**2
    total=np.sum(squared)
    loss=total/(2*m)
    print(f"w={w},b={b},loss={loss}") 
```

Model B is better than Model A. The cost function doesnt care about how many predication are exactly correct; it look all predications together. Model A makes two big errors and since we square the errors, big errors increases the cost much more than the small ones. In short , the models with lower cost is better model.

```python
import numpy as np
import matplotlib.pyplot as plt

x_train = np.array([1, 2, 3, 4, 5])
y_train = np.array([5, 7, 9, 11, 13])
m = x_train.shape[0]
b = 3

w_values = np.linspace(0, 4, 41)   
costs = []                         

for w in w_values:
    f = w * x_train + b
    errors = f - y_train
    squared = errors ** 2
    total = np.sum(squared)
    loss = total / (2 * m)
    costs.append(loss)  #turun sonucunu silinmeden deftere yazıyoruz.

plt.plot(w_values, costs)
plt.xlabel("w")
plt.ylabel("J(w)")
plt.show()
```
