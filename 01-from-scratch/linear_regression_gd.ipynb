```python
import numpy as np

x_train=np.array([1,2,3,4,5])
y_train=np.array([5,7,9,11,13])

m=x_train.shape[0]
b=3
w=0
alpha=0.05



```

```python
f=w*x_train+b
errors=f-y_train
dj_dw=np.sum(errors*x_train)/m

w=w-alpha*dj_dw

print("dj_dw:", dj_dw)
print("new w:", w)
```

```python
w = 0
num_iters = 20
alpha=0.15

for i in range(num_iters):
    f = w * x_train + b
    errors = f - y_train
    dj_dw = np.sum(errors * x_train) / m
    w = w - alpha * dj_dw
    print(f"step {i+1}: w = {w:.4f}")
```

```python
w = 0
alpha = 0.05
num_iters = 20
cost_history = []          # empty list to store costs

for i in range(num_iters):
    f = w * x_train + b
    errors = f - y_train
    dj_dw = np.sum(errors * x_train) / m
    w = w - alpha * dj_dw

    f_new = w * x_train + b
    errors_new = f_new - y_train
    loss = np.sum(errors_new ** 2) / (2 * m)
    cost_history.append(loss)

    print(f"step {i+1}: w = {w:.4f}, J = {loss:.4f}")
```

```python
import matplotlib.pyplot as plt

w = 0
# learning rate : controls how big each step is
alpha = 0.15
#how many steps we will take
num_iters = 20
#list to store the cost of each step
cost_history = []

for i in range(num_iters):
    f = w * x_train + b
    errors = f - y_train
    dj_dw = np.sum(errors * x_train) / m
    w = w - alpha * dj_dw

    f_new = w * x_train + b
    errors_new = f_new - y_train
    loss = np.sum(errors_new ** 2) / (2 * m)
    #store cost of this step
    cost_history.append(loss)
    print(f"step {i+1}: w = {w:.4f}, J = {loss:.4f}")
 

# label of x-axis: number of steps , label of y-axis: the 20 cost values we calculated
plt.plot(range(1, num_iters + 1), cost_history)
#label of x-axis
plt.xlabel("step")
#label of y-axis
plt.ylabel("cost J")
plt.show()
```

```python
import matplotlib.pyplot as plt

w = 0
b = 0
alpha = 0.05
num_iters = 1000
cost_history = []

for i in range(num_iters):
    # predictions and errors with the current w and b
    f = w * x_train + b
    errors = f - y_train

    # gradients: compute BOTH before updating
    dj_dw = np.sum(errors * x_train) / m
    dj_db =np.sum(errors)/m  

    # simultaneous update
    w = w - alpha * dj_dw
    b = b- alpha * dj_db 

    # cost with the new w and b
    f_new = w * x_train + b
    errors_new = f_new - y_train
    loss = np.sum(errors_new ** 2) / (2 * m)
    cost_history.append(loss)

    # print only every 100 steps
    if i % 100 == 0:
        print(f"step {i+1}: w = {w:.4f}, b = {b:.4f}, J = {loss:.4f}")

print(f"final: w = {w:.4f}, b = {b:.4f}")

plt.plot(range(1, num_iters + 1), cost_history)
plt.xlabel("step")
plt.ylabel("cost J")
plt.show()
```
