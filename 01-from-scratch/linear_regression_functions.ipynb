```python
import numpy as np
import matplotlib.pyplot as plt

x_train = np.array([1, 2, 3, 4, 5])
y_train = np.array([5, 7, 9, 11, 13])
```

```python
def predict(x,w,b):
    f=w*x+b
    return f
```

```python
def compute_cost(x,y,w,b):
    m=x.shape[0]
    f=predict(x,w,b)
    errors=f-y
    cost=np.sum(errors**2)/(2*m)
    return cost
```

```python
print(predict(x_train, 2, 3))                 
print(compute_cost(x_train, y_train, 1, 3))  
print(compute_cost(x_train, y_train, 2, 3))   
```

```python
def compute_gradient(x, y, w, b):
    m = x.shape[0]
    errors = predict(x,w,b) -y
    dj_dw =(1/m)*np.sum(errors*x)
    dj_db = (1/m)*np.sum(errors)
    return dj_dw, dj_db
```

```python
dj_dw, dj_db = compute_gradient(x_train, y_train, 0, 0)
print(dj_dw, dj_db)
dj_dw, dj_db = compute_gradient(x_train, y_train, 2, 3)
print(dj_dw, dj_db)
```

```python
def gradient_descent(x, y, w_init, b_init, alpha, num_iters):
    w = w_init
    b = b_init
    cost_history = []

    for i in range(num_iters):
        dj_dw, dj_db =compute_gradient(x,y,w,b)  
        w = w-alpha*dj_dw        
        b =b-alpha*dj_db
        cost = compute_cost(x,y,w,b)
        cost_history.append(cost)

    return w, b, cost_history
```

```python
w_final, b_final, cost_history = gradient_descent(x_train, y_train, 0, 0, 0.05, 1000)
print(f"w = {w_final:.4f}, b = {b_final:.4f}")   # w = 2.0001, b = 2.9995

plt.plot(cost_history)
plt.xlabel("step")
plt.ylabel("cost J")
plt.show()
```

```python
hours = np.array([1, 2, 4, 5])
scores = np.array([50, 55, 70, 75])

w_s, b_s, history_s = gradient_descent(hours, scores, 0, 0, 0.05, 2000)
print(f"w = {w_s:.4f}, b = {b_s:.4f}")
print(f"cost = {compute_cost(hours, scores, w_s, b_s):.4f}")
```

```python
cost_A = compute_cost(hours, scores, 5, 45)
cost_B = compute_cost(hours, scores, 8, 40)
cost_gd = compute_cost(hours, scores, w_s, b_s)

print(f"Model A:          J = {cost_A:.4f}")
print(f"Model B:          J = {cost_B:.4f}")
print(f"Gradient descent: J = {cost_gd:.4f}")
```
